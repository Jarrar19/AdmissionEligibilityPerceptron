```python
import sys
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Ensure project root is on path
sys.path.insert(0, str(Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()))
from src.config import FEATURE_COLS, TARGET_COL, RAW_DATA_PATH
from src.data_loader import load_data

sns.set_theme(style="whitegrid", palette="muted")
```

### 1. Dataset Overview & Schema Audit

```python
df = load_data()
print(f"Total Samples: {len(df)}")
print(f"Feature Count: {len(FEATURE_COLS)}")
print("\nMissing values count:")
print(df.isnull().sum())
df.head()
```

### 2. Descriptive Statistics

```python
df[FEATURE_COLS].describe().T.round(2)
```

### 3. Target Class Distribution (§7)

Reporting exact class counts and percentages to audit class balance before model training.

```python
counts = df[TARGET_COL].value_counts().sort_index()
pcts = df[TARGET_COL].value_counts(normalize=True).sort_index() * 100

print("Class Distribution:")
print(f"  Not Eligible (0): {counts[0]} ({pcts[0]:.1f}%)")
print(f"  Eligible (1):     {counts[1]} ({pcts[1]:.1f}%)")

plt.figure(figsize=(6, 4))
plt.bar(["Not Eligible (0)", "Eligible (1)"], counts.values, color=["#e74c3c", "#2ecc71"], width=0.5)
plt.title("Target Class Distribution")
plt.ylabel("Count")
plt.show()
```

### 4. Correlation Analysis

Examining Pearson correlation coefficients between academic predictors and admission eligibility.

```python
plt.figure(figsize=(10, 8))
corr = df[FEATURE_COLS + [TARGET_COL]].corr()
sns.heatmap(corr, annot=True, fmt=".2f", cmap="Blues", square=True)
plt.title("Correlation Matrix")
plt.show()
```

### 5. Written Findings & Observations (§7)

1. **Class Distribution**: The dataset contains 555 Eligible (55.5%) and 445 Not Eligible (44.5%) records. Class distribution is well-balanced, removing the need for synthetic oversampling (e.g. SMOTE). Stratified splitting will be applied to preserve this exact ratio.
2. **Strongest Predictors**: `CGPA` (correlation ~0.55), `Entrance_Score` (~0.58), and `Interview_Score` (~0.44) demonstrate the strongest positive linear association with admission eligibility.
3. **Negative Predictors**: `Previous_Backlogs` shows a strong negative association (~ -0.48), confirming that repeated course deficiencies substantially reduce eligibility probability.
4. **Class Overlap & Linearity**: Scatter plots of `CGPA` vs `Entrance_Score` reveal non-trivial overlap in the borderline transition zone (CGPA 7.0–8.0, Entrance 60–75). Because the Perceptron is a purely linear classifier, it cannot achieve 100% accuracy on this overlapping boundary — this is realistic and expected behavior.
